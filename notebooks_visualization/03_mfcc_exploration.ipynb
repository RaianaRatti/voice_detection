```python
from scipy.io import wavfile
from pathlib import Path
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
import sys

ROOT = Path().resolve().parents[0] # Jupyter equivalent

sys.path.append(str(ROOT))

from features.mfcc import extract
```

```python
sr1, empty_audio = wavfile.read(ROOT / "data" / "empty_audio.wav")
sr2, sample_audio = wavfile.read(ROOT / "data" / "sample_audio.wav")

empty_audio = empty_audio.astype(np.float32) / 32768.0
sample_audio = sample_audio.astype(np.float32) / 32768.0

mfcc1 = extract(empty_audio, sr1)
mfcc2 = extract(sample_audio, sr2)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
librosa.display.specshow(mfcc1, x_axis="time")
plt.title("Empty Audio MFCC")
plt.colorbar()

plt.subplot(1, 2, 2)
librosa.display.specshow(mfcc2, x_axis="time")
plt.title("Sample Audio MFCC")
plt.colorbar()

plt.tight_layout()
plt.show()
```
